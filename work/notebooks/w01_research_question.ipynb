## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**The question:** Given the observable signals we have (impressions, clicks, position, freshness, engagement), which pages should a content reviewer look at first when deciding what to refresh?

**Decision improved:** which pages go to the top of a limited-capacity review queue.

**Who acts on it:** a content editor or SEO reviewer with time to review only a fixed number of pages (e.g. the top 50) each cycle.

**Cost of a wrong call:** two-sided. If the score wrongly promotes a page that wasn't actually declining or losing opportunity, the reviewer wastes limited time on it. If the score misses a page that really is losing visibility, that page keeps declining unaddressed until the next review cycle — a real, measurable loss of search traffic.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]

n_pages = len(df)
n_declining = (df["trend_direction"] == "down").sum()
pct_declining = n_declining / n_pages * 100

print(f"Pages meeting the starter filter (impressions_90d>0, age>=90d): {n_pages:,}")
print(f"Pages currently trending down: {n_declining:,} ({pct_declining:.1f}% of eligible pages)")

# Verified results from outputs/model_report.md (starter pipeline, already run in this repo)
baseline_p50 = 0.240
model_p50 = 0.740
print(f"\nStarter baseline rule precision@50: {baseline_p50} -> ~{int(baseline_p50*50)} correct picks in the top 50")
print(f"Starter random forest precision@50: {model_p50} -> ~{int(model_p50*50)} correct picks in the top 50")
print("A learned ranking finds roughly three times as many real candidates as the hand-written rule, on this slice.")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What I can say:** this is a decision-support ranking. I can say a page shows *observed* signals of decline or opportunity, and that the ranking is *directional* — it points reviewers toward the pages most worth their limited time, based on evidence from the starter slice.

**What I can't say:** I can't claim that refreshing a page will *cause* it to recover — that would require a real experiment (e.g. before/after refresh with a control group), which this data doesn't give me. I also can't claim to predict Google's ranking algorithm, or that these results generalize beyond the ~30,000-row anonymized starter slice until I re-validate on the full warehouse release with proper client-holdout or time-aware splits.
